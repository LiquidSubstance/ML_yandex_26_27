```python
from pathlib import Path
import sys
import numpy as np
import pandas as pd

TASK = "task1"
cwd = Path.cwd()
candidates = [cwd, cwd / TASK, cwd / "student" / TASK,
              cwd / "regression_club" / "student" / TASK]
DATA_DIR = next((p for p in candidates if p.name == TASK and (p / "train.csv").is_file()), None)
if DATA_DIR is None:
    raise FileNotFoundError("Откройте ноутбук из папки задачи или корня комплекта")
sys.path.insert(0, str(DATA_DIR.parent))
from score import evaluate

train = pd.read_csv(DATA_DIR / "train.csv")
validation = pd.read_csv(DATA_DIR / "validation.csv")
test = pd.read_csv(DATA_DIR / "test.csv")
print("Строк:", len(train), len(validation), len(test))
```

```python
constant = float(train["target"].mean())
validation_predictions = np.full(len(validation), constant)
test_predictions = np.full(len(test), constant)
print(f"Константа: {constant:.6f}")
```

```python
validation[["id"]].assign(target=validation_predictions).to_csv(
    DATA_DIR / "baseline_validation.csv", index=False)
test[["id"]].assign(target=test_predictions).to_csv(
    DATA_DIR / "baseline_submission.csv", index=False)
result = evaluate(DATA_DIR / "validation.csv", DATA_DIR / "baseline_validation.csv",
                  DATA_DIR / "validation_calibration.json")
print(f"Validation MSE: {result['mse']:.6f}; score: {result['score']:.2f}/100")
print("Записаны baseline_validation.csv и baseline_submission.csv")
```
