```python
from pathlib import Path
import sys
import numpy as np
import pandas as pd
TASK = "task2"
cwd = Path.cwd()
candidates = [cwd, cwd / TASK, cwd / "student" / TASK, cwd / "logistic_club" / "student" / TASK]
DATA = next((p for p in candidates if p.name == TASK and (p.parent / "score_task1.py").is_file()), None)
if DATA is None:
    raise FileNotFoundError("Откройте ноутбук из папки задачи или корня комплекта")
sys.path.insert(0, str(DATA.parent))
```

```python
train = pd.read_csv(DATA / "train.csv")
constant = int(train["target"].mean() >= 0.5)
for split, filename in [("validation", "baseline_validation.csv"), ("test", "submission.csv")]:
    frame = pd.read_csv(DATA / f"{split}.csv", dtype={"id": str})
    frame[["id"]].assign(target=constant).to_csv(DATA / filename, index=False)
print("Записаны submission.csv и baseline_validation.csv")
```
