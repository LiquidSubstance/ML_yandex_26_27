```python
from pathlib import Path
import sys
import numpy as np
import pandas as pd
TASK = "task1"
cwd = Path.cwd()
candidates = [cwd, cwd / TASK, cwd / "student" / TASK, cwd / "logistic_club" / "student" / TASK]
DATA = next((p for p in candidates if p.name == TASK and (p.parent / "score_task1.py").is_file()), None)
if DATA is None:
    raise FileNotFoundError("Откройте ноутбук из папки задачи или корня комплекта")
sys.path.insert(0, str(DATA.parent))
```

```python
with np.load(DATA / "train.npz", allow_pickle=False) as a:
    train_ids = a["ids"]
    train_embeddings = np.zeros_like(a["X"])
for split, filename in [("validation", "baseline_validation.npz"), ("test", "baseline.npz")]:
    with np.load(DATA / f"{split}.npz", allow_pickle=False) as a:
        np.savez_compressed(DATA / filename, train_embeddings=train_embeddings,
                            train_ids=train_ids, eval_embeddings=np.zeros_like(a["X"]), eval_ids=a["ids"])
print("Записаны baseline.npz и baseline_validation.npz")
```

```python
from score_task1 import evaluate
result = evaluate(DATA / "baseline_validation.npz", DATA / "train.npz", DATA / "train_labels.csv",
                  DATA / "validation.npz", DATA / "validation_labels.csv", DATA / "validation_calibration.json")
print(result)
```
