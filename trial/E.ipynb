```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest
data = pd.read_csv("train.csv")
data["target"] = np.zeros(data.shape[0])
data.loc[data["rating"] == "A", "target"] = 1
data.loc[data["rating"] == "B", "target"] = 2
data.loc[data["rating"] == "C", "target"] = 3
data.loc[data["rating"] == "D", "target"] = 4
features = data.drop(["rating", "target", "id"], axis=1).columns
# for f in features:
#     (data.loc[data["target"] == 1, f]).hist(bins=20, color="green")
#     plt.xlabel(str(f))
#     plt.show()
#     (data.loc[data["target"] == 2, f]).hist(bins=20, color="lightgreen")
#     plt.xlabel(str(f))
#     plt.show()
#     (data.loc[data["target"] == 3, f]).hist(bins=20, color="red")
#     plt.xlabel(str(f))
#     plt.show()
#     (data.loc[data["target"] == 4, f]).hist(bins=20, color="darkred")
#     plt.xlabel(str(f))
#     plt.show()
train_features = ["income", "years_total_experience", "savings", "years_current_job" ]
X = np.array(data[train_features])
# best = SelectKBest(k=5, score_func=accuracy_score).fit_transform(X[features], data["target"])
# print(best)
model = LogisticRegression(C=30, max_iter=10000)
model.fit(X, data["target"])
test = pd.read_csv("test.csv")
prediction = model.predict(np.array(test[train_features]))
print(accuracy_score(data["target"], model.predict(X)))
test["rating"] = np.empty(test.shape[0], dtype=str)
test["prediction"] = prediction
test.loc[test["prediction"] == 1, "rating"] = "A"
test.loc[test["prediction"] == 2, "rating"] = "B"
test.loc[test["prediction"] == 3, "rating"] = "C"
test.loc[test["prediction"] == 4, "rating"] = "D"
print(test["rating"])
(test[["id", "rating"]]).to_csv("submission.csv", index=False)
```
